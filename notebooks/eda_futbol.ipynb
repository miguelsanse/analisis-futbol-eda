### Análisis de valores nulos

Antes de decidir qué columnas conservar, creo una función reutilizable que resume los 
valores nulos por columna, ordenados por porcentaje. Al ser una función, la aplico a 
las dos tablas sin duplicar código.

```python
def resumen_nulos(df):
    """
    Genera un resumen de valores nulos por columna de un DataFrame.

    Parámetros:
    - df (pd.DataFrame): DataFrame a analizar.

    Retorna:
    - pd.DataFrame: tabla con el número de nulos y su porcentaje por columna,
      ordenada de mayor a menor porcentaje de nulos.
    """
    total_nulos = df.isnull().sum()
    porcentaje_nulos = (total_nulos / len(df)) * 100

    resumen = pd.DataFrame({
        "nulos": total_nulos,
        "porcentaje": porcentaje_nulos.round(2)
    })

    resumen_ordenado = resumen.sort_values(by="porcentaje", ascending=False)
    return resumen_ordenado
```

```python
def asignar_temporada(fecha):
    """
    Asigna la temporada futbolística europea a una fecha dada.

    Una temporada va de agosto a mayo del año siguiente. Los partidos
    de agosto a diciembre pertenecen a la temporada que empieza ese año;
    los de enero a julio, a la temporada que empezó el año anterior.

    Parámetros:
    - fecha (datetime): fecha del partido.

    Retorna:
    - str: temporada en formato 'AAAA/AA' (por ejemplo, '2023/24').
    """
    anio = fecha.year
    mes = fecha.month

    if mes >= 8:
        anio_inicio = anio
    else:
        anio_inicio = anio - 1

    anio_fin = anio_inicio + 1
    return f"{anio_inicio}/{str(anio_fin)[-2:]}"


df_matches_limpio["temporada"] = df_matches_limpio["MatchDate"].apply(asignar_temporada)
```

### Detección de un error: la "temporada fantasma"

Al revisar el reparto de partidos por temporada, detecto una anomalía: aparece una 
temporada **"1999/00" con solo 19 partidos**, frente a los ~3.900 de una temporada normal.

La causa es el mes de corte. Al usar agosto (`mes >= 8`), los partidos de finales de 
julio —que algunas ligas como la Ligue 1 francesa ya disputan— quedan asignados a la 
temporada anterior, creando una temporada "fantasma" que en realidad no existe.

**Corrección:** cambio el mes de corte a julio (`mes >= 7`), de modo que esos partidos 
se asignen a la temporada que realmente arranca. A continuación redefino la función 
corregida y vuelvo a aplicarla a ambas tablas.

```python
def asignar_temporada(fecha):
    """
    Asigna la temporada futbolística europea a una fecha dada.

    Una temporada va de julio a mayo del año siguiente. Se toma julio como
    mes de corte porque algunas ligas (p. ej. la Ligue 1 francesa) disputan
    sus primeras jornadas a finales de julio. Los partidos de julio a
    diciembre pertenecen a la temporada que empieza ese año; los de enero
    a junio, a la temporada que empezó el año anterior.

    Parámetros:
    - fecha (datetime): fecha del partido o del registro de Elo.

    Retorna:
    - str: temporada en formato 'AAAA/AA' (por ejemplo, '2023/24').
    """
    anio = fecha.year
    mes = fecha.month

    if mes >= 7:
        anio_inicio = anio
    else:
        anio_inicio = anio - 1

    anio_fin = anio_inicio + 1
    return f"{anio_inicio}/{str(anio_fin)[-2:]}"
```

## 4. Variables derivadas para el análisis

Creo las variables que conectan la expectativa previa con el resultado real: la 
diferencia de Elo entre rivales (`diff_elo_inicio`), el total y la diferencia de goles, 
y los puntos obtenidos por el local. Estas variables alimentarán tanto el análisis 
estadístico como el dashboard.

```python
df_final["diff_elo_inicio"] = df_final["elo_inicio_home"] - df_final["elo_inicio_away"]

df_final["total_goles"] = df_final["FTHome"] + df_final["FTAway"]
df_final["diff_goles"] = df_final["FTHome"] - df_final["FTAway"]

def puntos_local(resultado):
    """
    Traduce el resultado de un partido a los puntos que obtuvo el equipo local.

    Sigue el sistema estándar del fútbol: 3 puntos por victoria,
    1 por empate, 0 por derrota.

    Parámetros:
    - resultado (str): 'H' (gana local), 'D' (empate), 'A' (gana visitante).

    Retorna:
    - int: puntos obtenidos por el equipo local (3, 1 o 0).
    """
    if resultado == "H":
        return 3
    elif resultado == "D":
        return 1
    else:
        return 0


df_final["puntos_local"] = df_final["FTResult"].apply(puntos_local)

print(df_final.shape)
df_final[["HomeTeam", "AwayTeam", "FTHome", "FTAway", "FTResult",
          "diff_elo_inicio", "total_goles", "diff_goles", "puntos_local"]].sample(8)
```

### Test de hipótesis: ¿existe la ventaja de jugar en casa?

El análisis descriptivo muestra que el local marca más goles de media (1,469) 
que el visitante (1,111). Para comprobar si esta diferencia es real y no fruto 
del azar, aplico una **prueba t de Student para muestras pareadas** (`ttest_rel`).

Uso la versión pareada porque los goles del local y del visitante proceden del 
mismo partido, es decir, son observaciones emparejadas, no dos grupos independientes.

- **H₀:** no hay diferencia entre la media de goles del local y del visitante.
- **H₁:** la media de goles del local es distinta a la del visitante.
- **Nivel de significación:** α = 0,05.

```python
from scipy import stats

goles_local = df_final["FTHome"]
goles_visitante = df_final["FTAway"]

print("Nulos en goles local:", goles_local.isnull().sum())
print("Nulos en goles visitante:", goles_visitante.isnull().sum())

t_stat, p_valor = stats.ttest_rel(goles_local, goles_visitante)

print(f"\nEstadístico t: {t_stat:.4f}")
print(f"p-valor: {p_valor}")
```

### Interpretación

El estadístico t es 66,05 y el p-valor es prácticamente cero (**p < 0,001**), 
muy por debajo de α = 0,05. Por tanto, **rechazo la hipótesis nula**.

Existe evidencia estadística muy fuerte de que el equipo local marca más goles 
que el visitante: la ventaja de jugar en casa es un efecto real, no azar.

**Matiz sobre el tamaño del efecto:** con una muestra tan grande (93.570 partidos), 
el test detecta como significativa casi cualquier diferencia. Por eso, más allá 
del p-valor, conviene mirar la magnitud real: la diferencia es de ~0,36 goles por 
partido a favor del local, que en el contexto del fútbol sí es relevante.

```python
from scipy import stats

datos_validos = df_final.dropna(subset=["diff_elo_inicio", "diff_goles"])

print("Partidos usados en el test:", len(datos_validos))
print("Partidos descartados por falta de Elo:", len(df_final) - len(datos_validos))

coef_pearson, p_valor = stats.pearsonr(
    datos_validos["diff_elo_inicio"],
    datos_validos["diff_goles"]
)

print(f"\nCoeficiente de correlación (r): {coef_pearson:.4f}")
print(f"p-valor: {p_valor}")
```

### Interpretación

La correlación de Pearson entre la ventaja de Elo inicial del local 
(`diff_elo_inicio`) y su diferencia de goles (`diff_goles`) es r = 0,30 
(p < 0,001), calculada sobre los 92.323 partidos con Elo disponible 
(se descartan 1.247 partidos, el 1,33%, por no tener Elo de referencia).

Se trata de una correlación positiva y moderada: cuanto mayor es la ventaja 
de Elo del equipo local al inicio de la temporada, mayor tiende a ser su 
diferencia de goles a favor. Esto confirma que el Elo previo aporta información 
real sobre el resultado y justifica la unión de ambas fuentes de datos.

Sobre la magnitud: un valor de 0,30 es el esperable en fútbol, un deporte de 
alta varianza donde el azar (penaltis, rebotes, expulsiones) pesa mucho en cada 
partido. El Elo capta la calidad estructural del equipo, pero no puede predecir 
lo que ocurre en 90 minutos concretos. Una correlación moderada refleja justamente 
eso: una tendencia clara, pero con amplio margen para la sorpresa.

Matices metodológicos:
- La significación es extrema (p < 0,001) por el gran tamaño de muestra, por lo 
  que el dato informativo es la magnitud del coeficiente (0,30), no el p-valor.
- Correlación no implica causalidad: el Elo no causa los goles; ambos son reflejo 
  de la calidad subyacente del equipo.

### Test de hipótesis: ¿difieren los goles entre ligas?

Comparo la media de goles totales por partido entre las 10 ligas mediante un 
**ANOVA de una vía** (`f_oneway`), la prueba adecuada para comparar las medias 
de más de dos grupos en una sola contrastación.

- **H₀:** todas las ligas tienen la misma media de goles por partido.
- **H₁:** al menos una liga tiene una media distinta.
- **Nivel de significación:** α = 0,05.

```python
from scipy import stats

grupos_goles = [
    grupo["total_goles"].values
    for nombre_liga, grupo in df_final.groupby("Division")
]

print("Número de grupos (ligas):", len(grupos_goles))

f_stat, p_valor = stats.f_oneway(*grupos_goles)

print(f"\nEstadístico F: {f_stat:.4f}")
print(f"p-valor: {p_valor}")
```

### Interpretación

El estadístico F es 132,27 con un p-valor prácticamente nulo (**p < 0,001**), 
por lo que **rechazo la hipótesis nula**: existen diferencias significativas en 
el número de goles entre ligas.

La tabla de medias revela el patrón concreto:
- La **Bundesliga (D1)** es la más goleadora (2,95 goles/partido), seguida de la 
  2. Bundesliga (2,79): el fútbol alemán destaca por su carácter ofensivo.
- Premier, Serie A y LaLiga se agrupan en torno a 2,67–2,72.
- Las segundas divisiones española (2,33) y francesa (2,35) son las menos goleadoras.

La diferencia entre la liga más y la menos goleadora supera el medio gol por 
partido, una brecha relevante en términos futbolísticos. El ANOVA confirma que 
estas diferencias no se deben al azar.

### Test de hipótesis: ¿depende el resultado de la liga?

Analizo si el resultado del partido (victoria local, empate o victoria visitante) 
está asociado a la liga, mediante una **prueba chi-cuadrado de independencia** 
(`chi2_contingency`) sobre la tabla de contingencia resultado × liga.

- **H₀:** el resultado y la liga son independientes (mismo reparto H/D/A en todas las ligas).
- **H₁:** el resultado y la liga están asociados.
- **Nivel de significación:** α = 0,05.

```python
from scipy import stats

chi2, p_valor, grados_libertad, frecuencias_esperadas = stats.chi2_contingency(tabla_contingencia)

print(f"Estadístico chi-cuadrado: {chi2:.4f}")
print(f"p-valor: {p_valor}")
print(f"Grados de libertad: {grados_libertad}")
```

### Interpretación

El estadístico chi-cuadrado es 323,23 (18 grados de libertad) con un p-valor 
prácticamente nulo (**p < 0,001**), por lo que **rechazo la hipótesis nula**: 
existe asociación estadística entre el resultado y la liga.

Ahora bien, la magnitud de esa asociación debe matizarse observando la tabla de 
porcentajes:
- La **victoria local es muy estable** entre ligas: oscila solo entre el 43,2% (I2) 
  y el 47,1% (SP1). La ventaja de jugar en casa es un fenómeno universal del fútbol 
  europeo, no dependiente de la liga.
- Las diferencias significativas se concentran en el **reparto empate / victoria 
  visitante**. Las segundas divisiones italiana (I2) y francesa (F2) destacan por su 
  alto porcentaje de empates (32,0% y 30,4%).

Este resultado es coherente con el ANOVA de goles: las mismas ligas menos goleadoras 
(I2, F2) son las que más empates registran, lo que refuerza su perfil de competiciones 
más cerradas y tácticas. Como en los tests anteriores, la significación extrema deriva 
del gran tamaño muestral, por lo que la lectura relevante es la magnitud y el patrón, 
no el p-valor.
